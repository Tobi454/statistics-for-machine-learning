# Statistics for Machine Learning

Statistics is important in Machine Learning because it helps me understand data, identify patterns, measure relationships, detect unusual values, and evaluate whether results are meaningful.

This section covers the statistics I studied as part of my Mathematics for Machine Learning foundation.

---

# 1. Descriptive Statistics

Descriptive statistics are used to summarize and understand a dataset.

The main things I looked at were:

* Mean
* Median
* Mode
* Minimum and maximum
* Range
* Variance
* Standard deviation
* Population vs sample statistics

## 1.1 Mean

The mean is the average of a set of values.

$$
\text{Mean} = \frac{\sum x_i}{n}
$$

For example:

```python
scores = [60, 70, 70, 75, 80, 85, 90]

mean = sum(scores) / len(scores)
print(mean)
```

Output:

```text
75.71428571428571
```

The mean is useful when I want to know the general center of my data.

However, the mean can be strongly affected by outliers.

---

## 1.2 Median

The median is the middle value after the data has been sorted.

Example:

```text
60, 70, 70, 75, 80, 85, 90
```

The median is:

```text
75
```

If there is an even number of values, the median is the average of the two middle values.

Example:

```text
60, 70, 75, 80
```

$$
\text{Median} = \frac{70 + 75}{2} = 72.5
$$

The median is generally more resistant to extreme values than the mean.

For example:

```text
20, 22, 23, 25, 100
```

The mean is:

```text
38
```

while the median is:

```text
23
```

The value `100` has a large effect on the mean but does not change the median very much.

---

## 1.3 Mode

The mode is the value that appears most frequently.

Example:

```text
60, 70, 70, 75, 80, 85, 90
```

The mode is:

```text
70
```

Python:

```python
import statistics

scores = [60, 70, 70, 75, 80, 85, 90]

print(statistics.mode(scores))
```

---

## 1.4 Minimum, Maximum and Range

The minimum is the smallest value.

The maximum is the largest value.

The range measures the difference between them.

$$
\text{Range} = \text{Maximum} - \text{Minimum}
$$

Example:

```text
60, 70, 70, 75, 80, 85, 90
```

Minimum:

```text
60
```

Maximum:

```text
90
```

Range:

```text
90 - 60 = 30
```

Python:

```python
minimum = min(scores)
maximum = max(scores)
data_range = maximum - minimum

print(minimum)
print(maximum)
print(data_range)
```

# 2. Variance and Standard Deviation

Variance and standard deviation measure how spread out the data is.

## 2.1 Variance

Variance measures the average squared distance of values from the mean.

For a population:

$$
\sigma^2 =
\frac{\sum (x_i-\mu)^2}{N}
$$

Where:

* $x_i$ = individual value
* $\mu$ = population mean
* $N$ = number of values
* $\sigma^2$ = population variance

A larger variance means the values are more spread out.

A smaller variance means the values are closer together.

---

## 2.2 Population vs Sample Variance

There is an important difference between population variance and sample variance.

Population variance divides by:

$$
N
$$

Sample variance divides by:

$$
n-1
$$

Sample variance uses Bessel's correction because the sample is being used to estimate the population.

In NumPy:

```python
import numpy as np

scores = [60, 70, 70, 75, 80, 85, 90]

population_variance = np.var(scores)
sample_variance = np.var(scores, ddof=1)

print(population_variance)
print(sample_variance)
```

---

## 2.3 Standard Deviation

Standard deviation is the square root of variance.

$$
\sigma = \sqrt{\sigma^2}
$$

Standard deviation is useful because it is expressed in the same units as the original data.

Python:

```python
population_std = np.std(scores)
sample_std = np.std(scores, ddof=1)

print(population_std)
print(sample_std)
```

In Machine Learning, standard deviation becomes especially important when working with feature scaling and standardization.

---

# 3. Percentiles, Quartiles and IQR

Percentiles describe the relative position of a value within a dataset.

For example:

* 25th percentile → 25% of the data is below this point
* 50th percentile → 50% of the data is below this point
* 75th percentile → 75% of the data is below this point

## 3.1 Quartiles

Quartiles divide data into four parts.

### Q1

Q1 is the 25th percentile.

### Q2

Q2 is the 50th percentile, which is also the median.

### Q3

Q3 is the 75th percentile.

So:

$$
Q1 = 25\%
$$

$$
Q2 = 50\%
$$

$$
Q3 = 75\%
$$

Python:

```python
import numpy as np

scores = [55, 58, 62, 64, 68, 70, 75, 78, 82, 90]

q1 = np.percentile(scores, 25)
q2 = np.percentile(scores, 50)
q3 = np.percentile(scores, 75)

print(q1)
print(q2)
print(q3)
```

---

## 3.2 Interquartile Range

The Interquartile Range (IQR) measures the spread of the middle 50% of the data.

$$
IQR = Q3 - Q1
$$

Python:

```python
iqr = q3 - q1

print(iqr)
```

The IQR is useful because it is less affected by extreme values than the range.

---

## 3.3 Detecting Outliers with IQR

A common method for detecting outliers uses the IQR.

Lower bound:

$$
Q1 - 1.5(IQR)
$$

Upper bound:

$$
Q3 + 1.5(IQR)
$$

Values below the lower bound or above the upper bound can be considered potential outliers.

Python:

```python
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

print(lower_bound)
print(upper_bound)
```

This is useful during Exploratory Data Analysis because unusual values can affect Machine Learning models.

---

# 4. Z-Scores and Standardization

A z-score tells me how far a value is from the mean in terms of standard deviations.

The formula is:

$$
z = \frac{x-\mu}{\sigma}
$$

Where:

* $x$ = value
* $\mu$ = mean
* $\sigma$ = standard deviation

## Example

Suppose:

```text
Mean = 70
Standard deviation = 10
```

For a score of 80:

$$
z = \frac{80-70}{10}
$$

$$
z = 1
$$

This means 80 is one standard deviation above the mean.

For a score of 50:

$$
z = \frac{50-70}{10}
$$

$$
z = -2
$$

This means 50 is two standard deviations below the mean.

---

## 4.1 Interpreting Z-Scores

A positive z-score means the value is above the mean.

A negative z-score means the value is below the mean.

A z-score close to zero means the value is close to the mean.

---

## 4.2 Standardization

Standardization transforms data so that the feature has approximately:

```text
Mean = 0
Standard deviation = 1
```

Using NumPy:

```python
import numpy as np

scores = np.array([60, 70, 70, 75, 80, 85, 90])

mean = np.mean(scores)
std = np.std(scores)

standardized = (scores - mean) / std

print(standardized)
```

---

## 4.3 StandardScaler

Scikit-learn provides `StandardScaler` for standardization.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scores = np.array([[60], [70], [70], [75], [80], [85], [90]])

scaled_scores = scaler.fit_transform(scores)

print(scaled_scores)
```

Standardization can be important for algorithms that are sensitive to the scale of features.

Examples include:

* K-Nearest Neighbors
* K-Means
* Support Vector Machines
* Logistic Regression
* Linear Regression
* Neural Networks

Tree-based models generally do not require feature scaling in the same way.

# 5. Correlation

Correlation measures the direction and strength of a linear relationship between two variables.

The correlation coefficient usually ranges from:

$$
-1 \leq r \leq 1
$$

## Positive Correlation

When one variable increases, the other tends to increase.

Example:

```text
Study Hours ↑
Score ↑
```

This would result in a positive correlation.

---

## Negative Correlation

When one variable increases, the other tends to decrease.

Example:

```text
Price ↑
Demand ↓
```

This would result in a negative correlation.

---

## No or Weak Linear Correlation

If two variables do not have a clear linear relationship, their correlation can be close to zero.

---

## 5.1 Pearson Correlation

Pearson correlation measures the strength and direction of a linear relationship.

The formula is:

$$
r =
\frac{Cov(X,Y)}
{\sigma_X\sigma_Y}
$$

The result is between -1 and 1.

Interpretation:

```text
r = 1    → perfect positive linear relationship
r = -1   → perfect negative linear relationship
r = 0    → no linear relationship
```

---

## 5.2 Correlation in Python

Using Pandas:

```python
df.corr(numeric_only=True)
```

For two specific columns:

```python
df["Study_Hours"].corr(df["Score"])
```

A scatter plot can also help visualize the relationship.

```python
import matplotlib.pyplot as plt

plt.scatter(df["Study_Hours"], df["Score"])

plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.title("Study Hours vs Score")

plt.show()
```

---

## 5.3 Correlation Does Not Mean Causation

This is an important statistical idea.

If two variables are correlated, it does not automatically mean that one causes the other.

Correlation tells me that a relationship exists.

It does not, by itself, prove why the relationship exists.

---

# 6. Covariance

Covariance measures how two variables change together.

The population covariance formula is:

$$
Cov(X,Y) =
\frac{\sum (X_i-\mu_X)(Y_i-\mu_Y)}{N}
$$

## Positive Covariance

If both variables tend to increase together, covariance is positive.

## Negative Covariance

If one variable tends to increase while the other decreases, covariance is negative.

## Covariance Near Zero

A covariance close to zero suggests little linear co-movement.

---

## 6.1 Covariance vs Correlation

Covariance and correlation are related, but they are not the same.

### Covariance

* Shows the direction of the relationship
* Depends on the scale of the variables
* Can have a wide range of values

### Correlation

* Shows direction and strength
* Is standardized
* Always falls between -1 and 1
* Is easier to interpret

Correlation can be thought of as a standardized form of covariance.

---

## 6.2 Covariance Matrix

When working with multiple variables, I can calculate a covariance matrix.

For example, if I have:

```text
Age
Study Hours
Score
```

the covariance matrix shows how each pair of variables changes together.

Python:

```python
df.cov(numeric_only=True)
```

Covariance matrices are important in Machine Learning and are used in techniques such as Principal Component Analysis (PCA).

# 7. Sampling

A population is the complete group I am interested in.

A sample is a smaller subset taken from that population.

Example:

```text
Population → Every student in a university

Sample → 500 students selected from the university
```

It is often impractical to collect data from an entire population, so samples are used.

---

## 7.1 Representative Sample

A good sample should reasonably represent the population.

If the population contains different groups, the sample should not heavily overrepresent one group unless that reflects the population.

---

## 7.2 Random Sampling

Random sampling gives members of a population a chance of being selected.

Python example:

```python
import numpy as np

population = np.arange(1, 101)

sample = np.random.choice(
    population,
    size=10,
    replace=False
)

print(sample)
```

---

## 7.3 Sampling Bias

Sampling bias occurs when the method used to collect the sample causes it to systematically differ from the population.

For example, if I want to understand the study habits of all students but only survey students inside the library, the sample may not represent everyone.

Sampling bias can lead to misleading conclusions.

---

## 7.4 Basic Sampling Methods

### Simple Random Sampling

Members are selected randomly from the population.

### Stratified Sampling

The population is divided into groups, called strata, and samples are taken from each group.

### Systematic Sampling

Members are selected using a fixed interval.

For example:

```text
Every 10th person
```

---

## 7.5 Sampling and Machine Learning

Sampling is useful when datasets are too large to work with completely during exploration or prototyping.

Sampling ideas are also related to how datasets are divided into:

* Training data
* Validation data
* Test data

However, dataset splitting and statistical sampling are not exactly the same thing.

# 8. Statistical Inference

Descriptive statistics describe the data I have.

Statistical inference goes further by using a sample to make conclusions about a larger population.

---

## 8.1 Population Parameter vs Sample Statistic

A **parameter** describes a population.

A **statistic** describes a sample.

Example:

```text
Population mean → parameter
Sample mean → statistic
```

A sample statistic can be used to estimate a population parameter.

---

## 8.2 Point Estimate

A point estimate is a single value used to estimate a population parameter.

For example, if I calculate the average score of a sample and get:

```text
72
```

I can use 72 as a point estimate of the population's mean score.

---

# 9. Confidence Intervals

A confidence interval gives a range of plausible values for a population parameter based on sample data.

Instead of saying:

```text
The population mean is exactly 72.
```

I might estimate a range such as:

```text
68 to 76
```

The range communicates uncertainty in the estimate.

A confidence interval is not simply the probability that the fixed population parameter is inside the calculated interval. The confidence level describes the long-run performance of the interval procedure.

---

## 9.1 Margin of Error

The margin of error describes how much uncertainty is associated with an estimate.

Conceptually:

$$
\text{Confidence Interval}
=
\text{Estimate} \pm \text{Margin of Error}
$$

A larger sample will generally give a more precise estimate, assuming the sampling method is appropriate.

---

# 10. Hypothesis Testing

Hypothesis testing is used to evaluate whether there is enough statistical evidence to reject a claim about a population.

There are two main hypotheses.

## Null Hypothesis

The null hypothesis is usually written as:

$$
H_0
$$

It represents the default assumption or no-effect/no-difference claim.

## Alternative Hypothesis

The alternative hypothesis is written as:

$$
H_1
$$

or:

$$
H_a
$$

It represents the claim that there is an effect, difference, or relationship.

---

## Example

Suppose I want to know whether a new teaching method changes average test scores.

I could define:

```text
H₀: The teaching method does not change the average score.

H₁: The teaching method changes the average score.
```

The statistical test then provides evidence that I can use to make a decision.

---

# 11. Test Statistic

A test statistic is a value calculated from the sample that measures how far the observed result is from what would be expected under the null hypothesis.

Different statistical tests use different test statistics.

The test statistic is then used to determine how unusual the observed result would be under the null hypothesis.

---

# 12. P-Value

The p-value is the probability, assuming the null hypothesis is true, of obtaining the observed result or something more extreme.

A small p-value means the observed result would be relatively unusual if the null hypothesis were true.

A common significance level is:

$$
\alpha = 0.05
$$

A common decision rule is:

```text
p < 0.05
→ Reject H₀

p ≥ 0.05
→ Do not reject H₀
```

It is better to say **"do not reject"** rather than **"accept"** the null hypothesis.

A p-value does not tell me the probability that the null hypothesis itself is true.

---

# 13. Type I and Type II Errors

Hypothesis testing can result in two important types of errors.

## Type I Error

A Type I error occurs when I reject a true null hypothesis.

In simple terms:

```text
False positive
```

Example:

I conclude that a treatment has an effect when it actually does not.

---

## Type II Error

A Type II error occurs when I fail to reject a false null hypothesis.

In simple terms:

```text
False negative
```

Example:

I conclude that there is not enough evidence of an effect when a real effect exists.

---

# 14. Statistical Significance vs Practical Importance

A statistically significant result is not automatically an important result in the real world.

For example, a very large dataset could produce a statistically significant difference that is extremely small in practice.

So when interpreting results, I should consider both:

```text
Statistical significance
+
Practical importance
```
